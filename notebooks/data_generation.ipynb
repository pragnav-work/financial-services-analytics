```python
import pandas as pd
import numpy as np
import uuid

#random seed
np.random.seed(42)

#date range
START_DATE = "2022-01-01"
END_DATE = "2026-12-31"

dates = pd.date_range(
    start=START_DATE,
    end=END_DATE,
    freq="MS"
)

#uuid generator
def generate_uuid():
    return str(uuid.uuid4())

print("Data generation environment ready!")
print("Start date:", START_DATE)
print("End date:", END_DATE)
print("Number of months:", len(dates))
```

```python
institutions = pd.DataFrame({
    "institution_id": [generate_uuid() for _ in range(5)],
    "institution_name": [
        "HDFC Financial Services",
        "ICICI Financial Group",
        "Axis Capital Services",
        "SBI Financial Solutions",
        "Kotak Investment Services"
    ],
    "location": [
        "Mumbai",
        "Chennai",
        "Bengaluru",
        "Delhi",
        "Hyderabad"
    ]
})

institutions
```

```python
product_data = [
    ("Growth Equity Fund", "Investment"),
    ("Balanced Mutual Fund", "Investment"),
    ("Fixed Income Fund", "Investment"),
    ("Retirement Investment Plan", "Investment"),
    ("Wealth Management Plan", "Investment"),
    ("Personal Loan", "Loan"),
    ("Home Loan", "Loan"),
    ("Business Loan", "Loan"),
]

products = pd.DataFrame([
    {
        "product_id": generate_uuid(),
        "institution_id": institution["institution_id"],
        "product_name": product_name,
        "product_type": product_type
    }
    for _, institution in institutions.iterrows()
    for product_name, product_type in product_data
])

products.head(10)
```

```python
print("Institutions:", len(institutions))
print("Products:", len(products))

print("\nProduct distribution:")
print(products["product_type"].value_counts())
```

```python
first_names = [
    "Aarav", "Arjun", "Rohan", "Rahul", "Aditya",
    "Priya", "Ananya", "Diya", "Sneha", "Kavya",
    "Ishaan", "Vikram", "Neha", "Meera", "Karan"
]

last_names = [
    "Sharma", "Patel", "Reddy", "Kumar", "Shah",
    "Iyer", "Nair", "Gupta", "Mehta", "Rao"
]

locations = [
    "Mumbai", "Chennai", "Bengaluru",
    "Delhi", "Hyderabad", "Pune", "Kolkata"
]

channels = [
    "Online", "Branch", "Referral", "Social Media"
]

customers = pd.DataFrame({
    "customer_id": [generate_uuid() for _ in range(5000)],
    "institution_id": np.random.choice(
        institutions["institution_id"], 5000
    ),
    "customer_name": [
        f"{np.random.choice(first_names)} {np.random.choice(last_names)}"
        for _ in range(5000)
    ],
    "location": np.random.choice(locations, 5000),
    "acquisition_date": pd.to_datetime(
        np.random.choice(
            pd.date_range(START_DATE, END_DATE),
            5000
        )
    ),
    "acquisition_channel": np.random.choice(
        channels,
        5000,
        p=[0.40, 0.25, 0.15, 0.20]
    )
})

customers.head()
```

```python
print("Total customers:", len(customers))
print("Unique customer IDs:", customers["customer_id"].nunique())
print("Null values:", customers.isnull().sum().sum())
```

```python
investment_customers = customers.sample(
    n=3000,
    random_state=42
).reset_index(drop=True)

investment_products = products[
    products["product_type"] == "Investment"
]

investment_accounts = pd.DataFrame({
    "account_id": [generate_uuid() for _ in range(3000)],
    "customer_id": investment_customers["customer_id"],
    "product_id": [
        np.random.choice(
            investment_products[
                investment_products["institution_id"] == row["institution_id"]
            ]["product_id"]
        )
        for _, row in investment_customers.iterrows()
    ],
    "opening_date": [
        row["acquisition_date"] + pd.Timedelta(
            days=np.random.randint(
                0,
                (pd.Timestamp(END_DATE) - row["acquisition_date"]).days + 1
            )
        )
        for _, row in investment_customers.iterrows()
    ],
    "account_status": np.random.choice(
        ["Active", "Closed"],
        3000,
        p=[0.90, 0.10]
    )
})

investment_accounts.head()
```

```python
print("Total investment accounts:", len(investment_accounts))
print("Unique account IDs:", investment_accounts["account_id"].nunique())
print("Null values:", investment_accounts.isnull().sum().sum())
print(investment_accounts["account_status"].value_counts())
```

```python
performance_data = []

month_ends = pd.date_range(
    start=START_DATE,
    end=END_DATE,
    freq="ME"
)

for _, account in investment_accounts[
    investment_accounts["account_status"] == "Active"
].iterrows():

    account_dates = month_ends[
        month_ends >= account["opening_date"]
    ]

    invested_amount = np.random.uniform(10000, 100000)

    market_value = invested_amount

    for date in account_dates:

        #monthly investment contribution
        invested_amount += np.random.uniform(1000, 5000)

        #market return with seasonal variation
        seasonal_effect = 0.004 if date.month in [10, 11, 12] else 0.001

        monthly_return = np.random.normal(
            0.008 + seasonal_effect,
            0.025
        )

        market_value *= (1 + monthly_return)

        performance_data.append({
            "performance_id": generate_uuid(),
            "account_id": account["account_id"],
            "snapshot_date": date,
            "invested_amount": round(invested_amount, 2),
            "market_value": round(max(market_value, 0), 2)
        })

investment_performance = pd.DataFrame(performance_data)

investment_performance.head()
```

```python
print("Total performance records:", len(investment_performance))
print("Unique performance IDs:", investment_performance["performance_id"].nunique())
print("Null values:", investment_performance.isnull().sum().sum())

print("\nDate range:")
print(investment_performance["snapshot_date"].min())
print(investment_performance["snapshot_date"].max())
```

```python
loan_customers = customers.sample(
    n=2500,
    random_state=24
).reset_index(drop=True)

loan_products = products[
    products["product_type"] == "Loan"
]

loan_accounts = pd.DataFrame({
    "loan_id": [generate_uuid() for _ in range(2500)],
    "customer_id": loan_customers["customer_id"],
    "product_id": [
        np.random.choice(
            loan_products[
                loan_products["institution_id"] == row["institution_id"]
            ]["product_id"]
        )
        for _, row in loan_customers.iterrows()
    ],
    "loan_amount": np.round(
        np.random.uniform(50000, 2000000, 2500), 2
    ),
    "loan_date": pd.to_datetime(
        np.random.choice(
            pd.date_range(START_DATE, END_DATE),
            2500
        )
    ),
    "loan_status": np.random.choice(
        ["Active", "Repaid", "Defaulted", "Closed"],
        2500,
        p=[0.55, 0.25, 0.12, 0.08]
    )
})

loan_accounts.head()
```

```python
survey_data = []

for _, customer in customers.iterrows():

    available_dates = pd.date_range(
        start=max(pd.Timestamp(START_DATE), customer["acquisition_date"]),
        end=pd.Timestamp(END_DATE)
    )

    survey_count = np.random.randint(
        1,
        min(3, len(available_dates)) + 1
    )

    survey_dates = np.random.choice(
        available_dates,
        survey_count,
        replace=False
    )

    for survey_date in survey_dates:

        survey_date = pd.Timestamp(survey_date)

        seasonal_adjustment = (
            -0.5 if survey_date.month in [3, 4] else 0.3
        )

        rating = int(np.clip(
            np.random.normal(7.5 + seasonal_adjustment, 2),
            0,
            10
        ))

        survey_data.append({
            "survey_id": generate_uuid(),
            "customer_id": customer["customer_id"],
            "survey_date": survey_date,
            "nps_rating": rating
        })

customer_surveys = pd.DataFrame(survey_data)

customer_surveys.head()
```

```python
expense_data = []

expense_dates = pd.date_range(
    start=START_DATE,
    end=END_DATE,
    freq="MS"
)

for _, institution in institutions.iterrows():

    for date in expense_dates:

        seasonal_factor = 1.4 if date.month in [1, 2, 3] else 1.0

        for expense_type in ["Marketing", "Sales"]:

            base_amount = (
                250000 if expense_type == "Marketing" else 150000
            )

            expense_amount = np.random.normal(
                base_amount * seasonal_factor,
                base_amount * 0.12
            )

            expense_data.append({
                "expense_id": generate_uuid(),
                "institution_id": institution["institution_id"],
                "expense_date": date,
                "expense_type": expense_type,
                "expense_amount": round(max(expense_amount, 0), 2)
            })

acquisition_expenses = pd.DataFrame(expense_data)

acquisition_expenses.head()
```

```python
import os

os.makedirs("../data", exist_ok=True)

datasets = {
    "institutions": institutions,
    "products": products,
    "customers": customers,
    "investment_accounts": investment_accounts,
    "investment_performance": investment_performance,
    "loan_accounts": loan_accounts,
    "customer_surveys": customer_surveys,
    "acquisition_expenses": acquisition_expenses
}

for name, df in datasets.items():

    df.to_csv(f"../data/{name}.csv", index=False)

    print(f"{name}.csv: {len(df)} records")
```
